```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten,Dense
from tensorflow.keras.utils import to_categorical
```

```python
(x_train,y_train),(x_test,y_test)=mnist.load_data()
```

```python
x_train , x_test = x_train/255.0 , x_test/255.0
```

```python
y_train_cat=to_categorical(y_train)
y_test_cat=to_categorical(y_test)
```

```python
print(f"Training images: {x_train.shape}")
print(f"Test images: {x_test.shape}")
```

Development of model

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model.compile(optimizer='sgd',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.summary()
```

Training and evaluation planning

```python
print("Training is in progress......")
```

```python
model.fit(x_train, y_train_cat, epochs=10, batch_size=32, validation_split=0.1)
loss, acc = model.evaluate(x_test, y_test_cat)
print()
print(f"Test Accuracy: {acc*100:.2f}%")
```

Prediction function

```python
import numpy as np
```

```python
def predict_digit(image_index):
    """Predict the digit of an image from the test set"""
    img = x_test[image_index]
    prediction = model.predict(img.reshape(1, 28, 28))
    predicted_label = np.argmax(prediction)
    actual_label = y_test[image_index]

    print(f"Predicted: {predicted_label}, Actual: {actual_label}")
    return predicted_label
```
